## Exploratory Data Analysis <br>
#Relationship between study hours and exam score<br>
Question-1<br>
Understanding how the no. of study hours directly affect a student's exam score<br>
Question-2<br>
Analysing if increased study hours result in increased exam scores



```python
plt.scatter(df['study_hours'], df['exam_score'])
plt.xlabel('Total Study Hours')
plt.ylabel('Exam Score')
plt.title('Study Hours vs Exam Score')
plt.show()
```

Observation- <br>
There is a positive correlation between the amount of hours studied and the exam grades in the scatter diagram. Students who take more time to study tend to perform better than those who do not. Although there are variations, it can be seen that more study time improves performance.
