Distribution of Exam Scores <br>

Understanding the overall spread and shape of exam scores across all students

```python
df["exam_score"].hist(bins=20)
plt.xlabel("Exam Score")
plt.ylabel("Number of Students")
plt.title("Distribution of Exam Scores")
plt.show()
df["exam_score"].describe()
```

Observation- <br>
The histogram shows that exam scores are right-skewed, with most students scoring between roughly 9 and 27 (25th–75th percentile). The mean (≈18.8) is higher than the median (≈18.0), which is a classic sign of right skew — a small number of high-scoring students are pulling the average upward. The scores range from 1 to about 64, so this dataset does not use a 0–100 scale. The target variable is not normally distributed, which should be considered when selecting and evaluating machine learning models.

Correlation Heatmap of All Numeric Features <br>
Question-1<br>
Understanding how strongly all numeric features are correlated with each other and with exam_score, not just individually against exam_score

```python
corr_matrix = df.corr(numeric_only=True)

plt.figure(figsize=(12, 8))
plt.imshow(corr_matrix, cmap="coolwarm", interpolation="none")
plt.colorbar(label="Correlation Coefficient")
plt.xticks(range(len(corr_matrix.columns)), corr_matrix.columns, rotation=90)
plt.yticks(range(len(corr_matrix.columns)), corr_matrix.columns)
plt.title("Correlation Heatmap of Numeric Features")
plt.tight_layout()
plt.show()

corr_matrix
```

Observation- <br>
The heatmap confirms most of the individual scatter-plot findings from earlier, but also reveals overlap between features that wasn't visible when looking at them one at a time.

Against exam_score: study_hours, focus_index, and productivity_score all show strong positive correlation (dark orange/red), matching the individual scatter plots. burnout_level shows a strong negative correlation (dark blue), also consistent with earlier findings. mental_health_score shows a moderate positive correlation, which wasn't tested individually before. Age, self_study_hours, online_classes_hours, gaming_hours, screen_time_hours, exercise_minutes, caffeine_intake_mg, part_time_job, and upcoming_deadline all show weak/near-zero correlation with exam_score, confirming they are unlikely to be strong individual predictors.

Between features (multicollinearity check): focus_index and productivity_score are very strongly correlated with each other (dark red), meaning they carry overlapping information — including both in the regression model may not add much beyond including one. burnout_level is strongly negatively correlated with sleep_hours, focus_index, productivity_score, and mental_health_score, suggesting these "wellbeing" features move together rather than independently. study_hours also shows some positive overlap with productivity_score.

This suggests that for the regression model, study_hours, focus_index (or productivity_score, not necessarily both), burnout_level, and mental_health_score are the strongest candidate predictors, while features like gaming_hours, screen_time_hours, and social_media_hours contribute little on their own and may be dropped or deprioritized.

Outlier Detection <br>
Outlier Detection using Boxplots <br>

Question-1
Checking for the presence of outliers in important numerical features.

```python
important_features = [
    "study_hours",
    "focus_index",
    "productivity_score",
    "burnout_level",
    "exam_score"
]

for col in important_features:
    plt.figure(figsize=(5,3))
    df.boxplot(column=col)
    plt.title(f"Boxplot of {col}")
    plt.show()
```

Observation-<br>
The boxplots help identify potential outliers in the numerical features. Although a few extreme values are present, they appear to be realistic observations rather than data entry errors. Therefore, these values are retained, as removing genuine observations could negatively affect the performance and generalizability of the machine learning model.

Correlation Ranking with Exam Score <br>

Question-1<br>
Identifying which numerical features have the strongest correlation with exam scores.

```python
corr = df.corr(numeric_only=True)

corr["exam_score"].sort_values(ascending=False)
```

Observation-

The correlation ranking highlights how strongly each numerical feature is related to the target variable (exam_score).

Among all the features, Productivity Score (r = 0.886) has the strongest positive correlation with exam score, making it the most influential predictor in this dataset. Focus Index (r = 0.750) also exhibits a strong positive relationship, followed by Mental Health Score (r = 0.547) and Study Hours (r = 0.513), both of which show moderate positive correlations. This indicates that students with higher productivity, better focus, good mental health, and increased study time generally tend to achieve higher exam scores.

Features such as Sleep Hours (r = 0.235) have only a weak positive relationship with exam performance, while Self Study Hours, Exercise Minutes, Online Classes Hours, and Age show little to no correlation with exam scores.

On the negative side, Burnout Level (r = -0.408) has the strongest negative correlation, suggesting that students experiencing higher burnout generally perform worse in examinations. Other features, including Upcoming Deadline, Part-time Job, Screen Time, Social Media Hours, Caffeine Intake, and Gaming Hours, display weak negative correlations and are unlikely to be strong individual predictors of exam performance.

Overall, the correlation analysis suggests that Productivity Score, Focus Index, Mental Health Score, Study Hours, and Burnout Level are the most promising features for predicting student exam scores and should be considered during the machine learning model-building phase.
